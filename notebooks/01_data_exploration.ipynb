```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_breast_cancer
```

```python
cancer = load_breast_cancer()
```

```python
print(cancer.keys())
```

```python
print(cancer.DESCR)
```

```python
df = pd.DataFrame(
    cancer.data,
    columns=cancer.feature_names
)
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["mean radius"], bins=20)

plt.title("Distribution of Mean Radius")
plt.xlabel("Mean Radius")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["mean texture"], bins=20)

plt.title("Distribution of Mean Texture")
plt.xlabel("Mean Texture")

plt.show()
```

```python
plt.figure(figsize=(12,10))

sns.heatmap(df.corr(), cmap="coolwarm")

plt.show()
```

- The dataset contains 569 samples and 30 numeric features.
- There are two target classes: malignant and benign.
- There are no missing values.
- Some features appear strongly correlated.
- The distributions of several features are skewed.
